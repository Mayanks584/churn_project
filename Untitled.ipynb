```python
print("Rows:", len(h))
print("Columns:", len(h.columns))
print("Duplicate rows:", h.duplicated().sum())
print("Unique users:", h["user_id"].nunique())
print("Repeated user rows:", h["user_id"].duplicated().sum())
```

```python
print(h["churn_status"].value_counts(dropna=False))

print("\nPercentage:")
print(
    h["churn_status"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)
```

```python
categorical_cols = h.select_dtypes(include=["object", "string"]).columns

for col in categorical_cols:
    print(f"\n===== {col} =====")
    print(h[col].value_counts(dropna=False).head(20))
```

```python
pd.crosstab(
    h["subscription_end_date"],
    h["churn_status"]
).head(20)
```

```python
pd.crosstab(
    h["subscription_end_date"],
    h["churn_status"],
    normalize="index"
).sort_values("Cancelled", ascending=False).head(20)
```
